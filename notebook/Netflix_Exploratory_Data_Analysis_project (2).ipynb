```python
df['director']=df['director'].fillna("Unknown")
df['cast']=df['cast'].fillna("Not Available")
df['country']=df['country'].fillna("Unknown")
```

# 1) Distribution of content types

```python
type_counts = df['type'].value_counts()
plt.figure(figsize=(10,6))
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', startangle=90, counterclock=False)
plt.title("Distribution of Content Types")
plt.legend()
plt.show()
```

# 2) Top 5 Actors on Netflix (Excluding "not avalable")

```python

# Store actor names
actors = []

# Get cast data from the DataFrame, excluding 'Not Available' entries
# This filters out NaN values that were filled with "Not Available"
cast_data = df[df['cast'] != "Not Available"]['cast']

# Split actor names and flatten the list
for cast_entry in cast_data:
    for actor in cast_entry.split(','):
        actors.append(actor.strip())

# Count actors
actor_series = pd.Series(actors)
# Filter out 'not avalable' before getting the top 10
actor_count = actor_series[actor_series != 'not avalable'].value_counts().head(5)
Colours=["blue","green","yellow","orange","red"]
# Plot
plt.figure(figsize=(10,5))
plt.barh(actor_count.index, actor_count.values, color=Colours)
plt.xlabel('Number of Movies / TV Shows')
plt.ylabel('Actor Name')
plt.title('Top 5 Actors on Netflix (Excluding "not avalable")') # Updated title
plt.gca().invert_yaxis()
plt.show()
```

# 3) Top 5 Directors on Netflix (Excluding "Not Avalable")

```python
# Count directors, excluding 'Not Avalable' (case-insensitive)
director = df['director']
director_count = director[director.str.lower() != 'not avalable'].value_counts().head(5)
colours=["indigo","darkblue","blue","skyblue","lightblue"]
# Plot
plt.figure(figsize=(10,5))
plt.barh(director_count.index, director_count.values, color=colours)
plt.xlabel('Number of Movies / TV Shows')
plt.ylabel('Director Name')
plt.title('Top 5 Directors on Netflix (Excluding "Not Avalable")')
plt.gca().invert_yaxis()
plt.show()
```

# 4) Movies and TV Shows Timeline by Release Year

```python
# Separate Movies and TV Shows
movies = df[df['type'] == 'Movie']
tvshows = df[df['type'] == 'TV Show']

# Count releases by year
movie_year = movies['release_year'].value_counts().sort_index()
tv_year = tvshows['release_year'].value_counts().sort_index()

# Plot line graph
plt.figure(figsize=(10,5))

plt.plot(movie_year.index,movie_year.values,label='Movies', linestyle="-.", linewidth=2)
plt.plot(tv_year.index,tv_year.values,label='TV Shows', linestyle="--", linewidth=2, color="magenta")

# Labels and title
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.title('Movies and TV Shows Timeline by Release Year')
plt.legend()
plt.show()
```

# 5) Netflix Release Year Distribution After 2011

```python
# Filter release years after 2011
years = df[df['release_year'] > 2011]['release_year']
colours=["violet","indigo","teal","darkblue","red","blue","Green","yellow","orange","lightblue"]

# Histogram
plt.figure(figsize=(10,5))
n, bins, patches = plt.hist(years, bins=10, edgecolor='black')

# Apply different colors to each bar
for i in range(len(patches)):
    patches[i].set_facecolor(colours[i % len(colours)])

plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.title('Netflix Release Year Distribution After 2011')

# Set x-axis ticks to be integers at bin edges and rotate for readability
plt.xticks(bins.astype(int))
plt.show()
```

# 6) Release Year Distribution After 2015

```python
# Filter release years after 2015
years = df[df['release_year'] > 2015]['release_year']

# Count titles per year
year_count = years.value_counts().sort_index()

# Pie chart
plt.figure(figsize=(10,9))
plt.pie(year_count.values,labels=year_count.index, autopct='%1.0f%%')
plt.title('Release Year Distribution After 2015')
plt.legend(loc="upper right")
plt.show()
```

# 7) Movies and TV Shows by Duration

```python
movies = df[df['type'] == 'Movie'].copy()
tvshows = df[df['type'] == 'TV Show'].copy()

# Extract numeric duration
movies.loc[:, 'duration_num'] = movies['duration'].str.extract(r'(\d+)').astype(float)
tvshows.loc[:, 'duration_num'] = tvshows['duration'].str.extract(r'(\d+)').astype(float)

# Scatter plot
plt.figure(figsize=(10,5))

# Movies
plt.scatter(movies['release_year'],movies['duration_num'],label='Movies', color="blue")
# TV Shows
plt.scatter(tvshows['release_year'],tvshows['duration_num'],label='TV Shows', color="magenta")
plt.xlabel('Release Year')
plt.ylabel('Duration (Minutes / Seasons)')
plt.title('Movies and TV Shows by Duration')
plt.legend()
plt.show()
```

# 8) Top 10 TV Shows Released in 2004


```python
# Filter TV Shows released in 2004
tv_2004 = df[(df['type'] == 'TV Show') & (df['release_year'] == 2004)]

# Take first 10 titles
top10 = tv_2004['title'].head(10)

# Create simple values for plotting
values = range(1, len(top10)+1)

# Plot
plt.figure(figsize=(10,5))
plt.barh(top10, values)

plt.xlabel('TV Shows')
plt.ylabel('Ranking')
plt.title('Top 10 TV Shows Released in 2004')

plt.gca().invert_yaxis()
plt.show()
```

# 9) Genre Distribution on Netflix

```python
# Count selected categories
kids_tv = df['listed_in'].str.contains("Kids' TV", na=False).sum()
family_movies = df['listed_in'].str.contains("Children & Family Movies", na=False).sum()
comedies = df['listed_in'].str.contains("Comedies", na=False).sum()
documentaries = df['listed_in'].str.contains("Documentaries", na=False).sum()

# Labels and values
labels = ["Kids' TV","Children & Family Movies","Comedies","Documentaries"]
values = [kids_tv,family_movies,comedies,documentaries]

# Pie chart
plt.figure(figsize=(10,10))
plt.pie(values,labels=labels,autopct='%1.0f%%')
plt.title('Genre Distribution on Netflix')
plt.legend()
plt.show()
```

# 10) Movie Duration Distribution More Than 2.5 Hours

```python
# Select only Movies
movies = df[df['type'] == 'Movie']

# Extract duration in minutes
duration = movies.loc[:,'duration'].str.extract('(\d+)').astype(float)

# Convert minutes to hours
hours = duration / 60

# Filter movies more than 2.5 hours
long_movies = hours[hours[0] > 2.5]

# Histogram
plt.figure(figsize=(8,5))
plt.hist(long_movies, bins=10, color="pink", edgecolor="black")
plt.xlabel('Duration (Hours)')
plt.ylabel('Number of Movies')
plt.title('Movie Duration Distribution More Than 2.5 Hours')
plt.show()
```
